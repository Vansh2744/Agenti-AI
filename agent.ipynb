```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
```

```python
load_dotenv()

llm = ChatGroq(model="llama-3.1-8b-instant")

@tool
def get_weather(city:str)->str:
    """Get weather for the city"""

    return f"Weather of {city} is sunny"

@tool
def get_top_news(city:str)->str:
    """Get top news for the city"""

    return f"Top news of the city is that there is a big accident happen in {city}"

agent = create_agent(
    model=llm,
    tools=[get_weather, get_top_news],
    system_prompt="You are an helpful assistant"
)
agent
```

```python
res = agent.invoke(
    {"messages": [{"role": "user", "content": "What is the weather of bangalore"}]},
)

res["messages"][-1].content
```
